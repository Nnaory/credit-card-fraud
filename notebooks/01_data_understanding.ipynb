### 3.2 Column Identification
This subsection classifies each dataset column into distinct feature roles based on semantic domain meaning, measurement scale, and data type.


```python
df_train.info()
```

**Insight 3.2** Based on semantic domain characteristics, data types, and distribution properties, the columns are formally categorized into the following distinct feature groupings:

- **Identifier (Column Index 0)**:
  - `id`: Unique record identifier for each transaction observation.
- **Numerical Features (Column Indices 1–13, excluding cyclical features)**:
  - `amount_usd`: Continuous numerical feature denoting transaction value in US dollars.
  - `hours_since_last_txn`: Continuous numerical feature measuring elapsed hours since preceding transaction.
  - `txn_count_last_24h`: Discrete numerical feature measuring transaction frequency within the past 24 hours.
  - `distance_from_home_km`: Continuous numerical feature representing geographic distance from customer residence.
  - `card_age_months`: Discrete numerical feature representing the age of the card in months.
  - `customer_age`: Discrete numerical feature representing cardholder age in years.
  - `account_balance_usd`: Continuous numerical feature indicating available account balance in US dollars.
  - `cvv_retry_count`: Discrete numerical feature tracking retry attempts for CVV validation.
  - `velocity_score`: Continuous numerical feature quantifying transaction velocity dynamics.
  - `merchant_risk_score`: Continuous numerical feature reflecting risk assessment score for the merchant.
  - `prior_disputes`: Discrete numerical feature recording historical dispute occurrences.
- **Categorical Features (Column Indices 14–18)**:
  - `merchant_category`: Categorical string specifying merchant industry sector (12 distinct categories).
  - `card_type`: Categorical string denoting payment card brand or tier (5 distinct categories).
  - `auth_method`: Categorical string representing verification method used (5 distinct categories).
  - `channel`: Categorical string designating transaction acquisition channel (5 distinct categories).
  - `device_type`: Categorical string detailing client hardware/platform (8 distinct categories).
- **Boolean Features (Column Indices 19–24)**:
  - `is_foreign_transaction`: Binary flag indicating whether the transaction occurred outside the home country.
  - `is_new_merchant`: Binary flag indicating whether the transaction was completed with an unfamiliar merchant.
  - `used_vpn`: Binary flag indicating virtual private network (VPN) usage during transaction routing.
  - `ip_country_mismatch`: Binary flag indicating inconsistency between IP geolocation and home country.
  - `billing_shipping_mismatch`: Binary flag indicating discrepancy between billing and shipping locations.
  - `is_ai_generated_scam_attempt`: Binary flag indicating potential automated or synthetic scam indicators.
- **Cyclical Temporal Features (Column Indices 10–11)**:
  - `time_of_day_hour`: Integer hour feature bounded within [0, 23], reflecting a 24-hour periodic daily cycle.
  - `day_of_week`: Integer day feature bounded within [0, 6], reflecting a 7-day periodic weekly cycle.
- **Target Feature (Column Index 25)**:
  - `is_fraud`: Ground-truth binary target label (0.0 for non-fraudulent, 1.0 for fraudulent transactions).

Following this categorization, subsequent analysis focuses on targeted investigations of domain-critical features requiring specialized preprocessing considerations.


### 3.3 Column Investigation
This subsection conducts a targeted inspection of domain-critical cyclical and boolean features to examine boundary values and evaluate preprocessing requirements.


```python
CYCLICAL_FEATURES = ["day_of_week", "time_of_day_hour"]
BOOLEAN_FEATURES = ["is_foreign_transaction", "is_new_merchant", "used_vpn", "ip_country_mismatch", "billing_shipping_mismatch", "is_ai_generated_scam_attempt"]
```

```python
for column in CYCLICAL_FEATURES:
    print("Column: ", column)
    print("Unique Values: ", df_train[column].nunique())
    print("Max Values: ", df_train[column].max())
    print("Min Values: ", df_train[column].min())
    print(" ")
```

**Insight 3.3** The empirical investigation of temporal attributes confirms that `time_of_day_hour` is strictly bounded between 0 and 23 (24 discrete hours) and `day_of_week` is strictly bounded between 0 and 6 (7 discrete days). While represented as discrete integers, linear treatment of these features distorts circular continuity—hour 23 naturally wraps around to hour 0, and day 6 transitions into day 0. Consequently, cyclical trigonometric transformations (sine and cosine encodings) should be applied during feature engineering to preserve periodic distance relationships without introducing artificial boundary penalties.
